```python
import numpy as np
import pandas as pd

df = pd.DataFrame({'cgpa': np.random.randint(3, 10, 100), 'iq': np.random.randint(70, 140, 100), 'placed': np.random.randint(0,1, 100)})
df
```

```python
epochs = 1
n = 0.01

weights = np.ones(df.shape[1]) # [ Wintercept, Wcgpa, Wiq ]

for i in range(epochs):
	randn = np.random.randint(df.shape[0])
	rand_row = df.loc[randn].values    #[cgpa, iq, placed]

	xi = np.insert(rand_row[:-1], 0, 1) #[1, cgpa, iq]
	wixi = weights @ xi

	y = rand_row[-1]
	if y == 0 and wixi > 0: # if negative but predicted positive
		weights = weights - n * xi

	if y == 1 and wixi < 0: # if positve but predicted negative
		weights = weights + n * xi
	
print(weights)
```
