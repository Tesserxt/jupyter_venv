```python
from matplotlib import pyplot as plt
import numpy as np
import pandas as pd
from sklearn.datasets import make_classification, make_regression

#pd.DataFrame({'cgpa': np.random.randint(3, 10, 100), 'iq': np.random.randint(70, 140, 100), 'placed': np.random.randint(0,1, 100)})


X, y = make_classification(n_samples=100, n_features=2, n_informative=1,n_redundant=0,
						   n_classes=2, n_clusters_per_class=1, random_state=41,hypercube=False,class_sep=20)
```

```python
df = pd.DataFrame(np.append(X,y.reshape(-1,1),axis=1))
```

```python
epochs = 1000
n = 0.1

weights = np.ones(df.shape[1]) # [ Wintercept, Wcgpa, Wiq ]

for i in range(epochs):
	randn = np.random.randint(df.shape[0])
	rand_row = df.loc[randn].values    #[cgpa, iq, placed]

	xi = np.insert(rand_row[:-1], 0, 1) #[1, cgpa, iq]
	wixi = weights @ xi

	yhat = rand_row[-1]
	if yhat == 0 and wixi > 0: # if negative but predicted positive
		weights = weights - n * xi

	if yhat == 1 and wixi < 0: # if positve but predicted negative
		weights = weights + n * xi
	
print(weights)
```

```python
def perceptron(df, epochs=1000, lr=0.01):

	def step(value):
		return 1 if value > 0 else 0
	
	def sigmoid(value):
		return 1/(1+np.exp(-value))
	
	
	X =  np.insert(df.iloc[:,0:-1],0,1,axis=1)
	weights = np.ones(X.shape[1])

	for i in range(epochs):
		randn = np.random.randint(df.shape[0])
		yhat = sigmoid(X[randn] @ weights)
		y = df.iloc[randn, -1]
			
		weights = weights + lr * (y - yhat) * X[randn]

	return weights	# Bo, B1, B2...

weights = perceptron(df, epochs=1000)
intercept_ = weights[0]
coef_ = weights[1:]
print(intercept_, coef_)

m = -(coef_[0]/coef_[1])
b = -(intercept_/coef_[1])

x_input = np.linspace(-3,3,100)
y_input = m*x_input + b

plt.scatter(X[:, 0], X[:,1], c=y, cmap='winter', s=100)
plt.plot(x_input, y_input, color='red')
plt.ylim(-3,2)
```

```python
class LogisticRegression: # usign Batch Gradient Descent
    
	def __init__(self, epochs=1000, learning_rate=0.01):
		self.coef_ = None
		self.intercept_ = None
		self.epochs = epochs
		self.learning_rate = learning_rate

	def sigmoid(value):
		return 1/(1+np.exp(-value))

	def fit(self, X_train, y_train):
		X =  np.insert(X_train,0,1,axis=1)
		W = np.ones(X.shape[1])

		for _ in range(self.epochs):	
			yhat = LogisticRegression.sigmoid(X @ W)
			W = W + self.learning_rate * (((y_train - yhat) @ X) / X.shape[0])
		
		self.intercept_ = W[0]
		self.coef_ = W[1:]

LOR = LogisticRegression()
LOR.fit(X, y)

LOR.intercept_, LOR.coef_
```
