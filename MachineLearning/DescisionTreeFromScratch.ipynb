```python

def gini(y):
	total = y.value_counts().sum()
	nYes = y.value_counts()[1]
	nNo = y.value_counts()[0]

	prob_true = (nYes/total)**2
	prob_false = (nNo/total)**2

	return 1 - np.sum(prob_true + prob_false)

def gini(y):
	'''
	1 - ∑(p²)
	'''
	return 1 - np.sum(y.value_counts(normalize=True) ** 2)

def weighted_gini(y_left, y_right):
	total = len(y_left) + len(y_right)
	gini_left = gini(y_left)
	gini_right = gini(y_right)
	return (len(y_left)/total) * gini_left + (len(y_right)/total) * gini_right
```

```python
def best_split(X, y):
	best_gini = 1
	best_feature = None
	best_threshold = None
	for feature in X.columns:
		#feature = 'lovesSoda'
		values = X[feature].unique()
		
		for val in values:
			left_idx = X[feature] <= val #  Does not lovesPopcorn
			right_idx = ~left_idx		# 	lovesPopcorn

			#target variable
			y_left = y[left_idx] # Does not lovesPopcorn(0) and lovesCoolAsIce(0,1)
			y_right = y[right_idx] # lovesPopcorn(1) and lovesCoolAsIce(0,1)
			

			if len(y_left) == 0 or len(y_right) == 0: 
				continue
			
			gini_score = weighted_gini(y_left, y_right)

			if gini_score < best_gini:
				best_feature = feature
				best_gini = gini_score
				best_threshold = val
			
			#print(feature, val, np.round(gini_score, 2), np.round(best_gini,2), best_feature, best_threshold)
	
	return best_feature, best_threshold
		
```

```python
class Node:
	def __init__(self, feature=None, threshold=None, left=None, right=None, value=None):
		self.feature = feature
		self.threshold = threshold
		self.left = left
		self.right = right
		self.value = value
	
	def is_leaf(self):
		  return self.value is not None
	

```

```python
def build_tree(X, y, depth=0, max_depth=5):
	# Stopping criteria
	if len(y.unique()) == 1 or depth == max_depth:
		return Node(value=y.mode()[0])  # most common class

	feat, thresh = best_split(X, y)

	if feat is None: #triggers when all features have same value. Hence, no best feature to select
		return Node(value=y.mode()[0])

	left_idx = X[feat] <= thresh
	right_idx = ~left_idx

	left = build_tree(X[left_idx], y[left_idx], depth+1, max_depth)
	right = build_tree(X[right_idx], y[right_idx], depth+1, max_depth)

	return Node(feature=feat, threshold=thresh, left=left, right=right)
```

```python
def predict_one(X_one, node):
	if node.is_leaf():
		return node.value
	
	
	if X_one[node.feature] <= node.threshold: # if does not lovesSoda (left-no)
		return predict_one(X_one, node.left)
	
	else:								# if lovesSoda (right-yes)
		return predict_one(X_one, node.right)

def predict(X, tree):
	return X.apply(lambda row: predict_one(row, tree), axis=1)

```

```python
# Example
from sklearn.datasets import load_iris
import pandas as pd
from sklearn.model_selection import train_test_split

data = load_iris()
X = pd.DataFrame(data.data, columns=data.feature_names)
y = pd.Series(data.target)

X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42, test_size=0.2)
tree = build_tree(X_train, y_train, max_depth=2)
preds = predict(X_test, tree)

from sklearn.metrics import accuracy_score

accuracy_score(preds, y_test)
```
